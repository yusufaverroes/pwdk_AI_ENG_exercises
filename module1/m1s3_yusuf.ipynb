## Soal 5

```python
jarak_mobil_A_B=120
A_velocity= 60
B_velocity= 40
time_crash = (jarak_mobil_A_B / (A_velocity + B_velocity))+9 ##hour
hours = int(time_crash//1) #hour
minutes = round((time_crash-hours)*60)#minutes
print(f"Mobil A dan B akan bertabrakan pukul {hours}:{minutes} WIB")
```

## Soal 6

```python
jumlah_apel = int(input("Masukan jumlah apel: "))
jumlah_jeruk = int(input("Masukan jumlah jeruk: "))
jumlah_anggur = int(input("Masukan jumlah anggur: "))

harga_apel = 10000
harga_jeruk = 15000
harga_anggur = 20000

total_apel = jumlah_apel * harga_apel
total_jeruk = jumlah_jeruk * harga_jeruk
total_anggur = jumlah_anggur * harga_anggur
total_belanja = total_apel + total_jeruk + total_anggur

detail_belanja = f"""
Detail Belanja:
apel: {jumlah_apel} x {harga_apel} = {total_apel}
jeruk: {jumlah_jeruk} x {harga_jeruk} = {total_jeruk}
anggur: {jumlah_anggur} x {harga_anggur} = {total_anggur}
Total : {total_belanja}
"""
print(detail_belanja)
```

## Soal 3

```python
data_buah =[("apel",20,10000),
            ("jeruk",15,15000),
            ("anggur",25,20000), 
            ("nanas",10,25000)] #nama, stok, harga

nama,stock,harga = 0,1,2
data_buah_dibeli = []
exit_program = False
while not exit_program:
    panduan_program = """"
    Selamat Datang di Pasar Buah
    List Menu:
    1. Menampilkan daftar buah
    2. Menambah buah
    3. Menghapus buah
    4. Membeli buah
    5. Exit Program
    Masukan angka Menu yang ingin dijalankan: 
    """
    menu = input(panduan_program)
    if menu == "3":
        index_buah = input("Masukan index buah yang ingin dihapus: ")
        del data_buah[int(index_buah)]

    elif menu == "2":
        nama_buah = input("Masukan nama buah: ")
        stok_buah = int(input("Masukan stok buah: "))
        harga_buah = int(input("Masukan harga buah: "))
        data_buah.append((nama_buah,stok_buah,harga_buah))
    elif menu == "1":
         print("")
    elif menu == "4":
        mau_beli=True
        while mau_beli:
            index_buah = int(input("Masukan index buah yang ingin dibeli: "))
            jumlah_beli = int(input("Masukan jumlah buah yang ingin dibeli: "))
            if jumlah_beli > data_buah[index_buah][stock]:
                print(f"Stok tidak cukup, stok {data_buah[index_buah][nama]} tinggal {data_buah[index_buah][stock]}")
            else:
                print("isi cart : ")
                print("nama   | qty | harga")
                data_buah_dibeli.append((data_buah[index_buah][nama],jumlah_beli,data_buah[index_buah][harga]))
                data_buah[index_buah] = (data_buah[index_buah][nama],data_buah[index_buah][stock]-jumlah_beli,data_buah[index_buah][harga])
                for i in range(len(data_buah_dibeli)):
                    print(f"{data_buah_dibeli[i][nama]}   | {data_buah_dibeli[i][stock]}   | {data_buah_dibeli[i][harga]}")
                while True:
                    lanjut_beli = input("Apakah ingin membeli buah lain? (ya/tidak): ")
                    if lanjut_beli == "ya":
                        break
                    elif lanjut_beli == "tidak":
                        mau_beli = False
                        total_harga = sum(item[stock] * item[harga] for item in data_buah_dibeli)
                        print(f"Total harga yang harus dibayar adalah {total_harga}")
                        cash_in = 0 
                        while cash_in < total_harga:
                            cash_in += int(input("Masukan uang pembayaran: "))
                            if cash_in < total_harga:
                                print(f"Uang yang anda masukan kurang, sisa pembayaran adalah {total_harga-cash_in}")
                            elif cash_in >= total_harga:
                                print(f""" Terimakasih
                                      uang kembalian anda {cash_in-total_harga} """)
                        break
                    else:
                        print("Input tidak valid, masukan ya atau tidak")
            
    elif menu == "5":
        exit_program = True            
    else:
         print("Menu tidak tersedia, silahkan masukan angka 1-5")
    print("""
        Daftar Buah:
              
        index | Nama Buah | Stok | Harga
              """ )
    for i in range(len(data_buah)):
            print(f"{i}     | {data_buah[i][nama]}    | {data_buah[i][stock]}   | {data_buah[i][harga]}")
```

### Dictionary

```python
data_buah_dict = {
    "apel":{"stock":20,"harga":10000},
    "jeruk":{"stock":15,"harga":15000},
    "anggur":{"stock":25,"harga":20000},
    "nanas":{"stock":10,"harga":25000}
}


data_buah_dibeli = {}
exit_program = False
while not exit_program:
    panduan_program = """"
    Selamat Datang di Pasar Buah
    List Menu:
    1. Menampilkan daftar buah
    2. Menambah buah
    3. Menghapus buah
    4. Membeli buah
    5. Exit Program
    Masukan angka Menu yang ingin dijalankan: 
    """
    menu = input(panduan_program)
    if menu == "3":
        nama_buah = input("Masukan nama buah yang ingin dihapus: ")
        del data_buah_dict[nama_buah]

    elif menu == "2":
        nama_buah = input("Masukan nama buah: ")
        stok_buah = int(input("Masukan stok buah: "))
        harga_buah = int(input("Masukan harga buah: "))
        data_buah_dict.update({nama_buah:{"stock":stok_buah,"harga":harga_buah}})
    elif menu == "1":
         print("")
    elif menu == "4":
        mau_beli=True
        while mau_beli:
            nama_buah = (input("Masukan nama buah yang ingin dibeli: "))
            jumlah_beli = int(input("Masukan jumlah buah yang ingin dibeli: "))
            if jumlah_beli > data_buah_dict[nama_buah]["stock"]:
                print(f"Stok tidak cukup, stok {nama_buah} tinggal {data_buah_dict[nama_buah]["stock"]}")
            else:
                print("isi cart : ")
                print("nama   | qty | harga")
                data_buah_dibeli.update({nama_buah:{"qty":jumlah_beli,"harga":data_buah_dict[nama_buah]["harga"]}})
                data_buah_dict[nama_buah] = ({"stock":data_buah_dict[nama_buah]["stock"]-jumlah_beli,"harga":data_buah_dict[nama_buah]["harga"]})
                for key in data_buah_dibeli:
                    print(f"{key}   | {data_buah_dibeli[key]["qty"]}   | {data_buah_dibeli[key]["harga"]}")
                while True:
                    lanjut_beli = input("Apakah ingin membeli buah lain? (ya/tidak): ")
                    if lanjut_beli == "ya":
                        break
                    elif lanjut_beli == "tidak":
                        mau_beli = False
                        total_harga = sum(item["qty"] * item["harga"] for item in data_buah_dibeli.values())
                        print(f"Total harga yang harus dibayar adalah {total_harga}")
                        cash_in = 0 
                        while cash_in < total_harga:
                            cash_in += int(input("Masukan uang pembayaran: "))
                            if cash_in < total_harga:
                                print(f"Uang yang anda masukan kurang, sisa pembayaran adalah {total_harga-cash_in}")
                            elif cash_in >= total_harga:
                                print(f""" Terimakasih
                                      uang kembalian anda {cash_in-total_harga} """)
                        break
                    else:
                        print("Input tidak valid, masukan ya atau tidak")
            
    elif menu == "5":
        exit_program = True            
    else:
         print("Menu tidak tersedia, silahkan masukan angka 1-5")
    print("""
    Daftar Buah:
              
    Nama Buah | Stok | Harga
              """ )
    for i in data_buah_dict:
            print(f"{i}    | {data_buah_dict[i]['stock']}   | {data_buah_dict[i]['harga']}")
```

```python
film_kesukaan_anda = list(input("Masukan 5 film kesukaan anda: ").split(","))
film_kesukaan_teman_anda = list(input("Masukan 5 film kesukaan teman anda: ").split(","))
film_kesukaan_anda_set = set(film_kesukaan_anda[:5])
film_kesukaan_teman_anda_set = set(film_kesukaan_teman_anda[:5])
kesamaan_film = film_kesukaan_anda_set.intersection(film_kesukaan_teman_anda_set)
print(f"kesukaan film anda dan teman anda yang sama sebesar {len(kesamaan_film)/5*100}%")
```

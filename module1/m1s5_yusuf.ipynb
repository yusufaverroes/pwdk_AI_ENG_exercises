## Soal 1

```python
bentuk = 0 
while bentuk != 1 and bentuk != 2:
    bentuk = int(input("masukan bentuk '1' atau '2': "))

if bentuk == 1:
    for baris in range(9):
        jumlah_bintang=0
        if baris<=4:
            jumlah_bintang=5-baris
        else: 
            jumlah_bintang= baris-3
        for kolom in range(jumlah_bintang):
            print("* ", end="")
        print("\n")
elif bentuk == 2:
    for baris in range(9):
        jumlah_spasi=0
        jumlah_bintang=0
        if baris<=4:
            jumlah_bintang=baris+1
            jumlah_spasi=4-baris
        else: 
            jumlah_bintang=9-baris
            jumlah_spasi=baris-4
        for kolom in range(jumlah_bintang+jumlah_spasi):
            if kolom<jumlah_spasi:
                print(" ", end="")
            else: print("* ", end="")
        print("\n")
```

## Soal 2

```python
stok_jeruk = 7
stok_anggur = 6
stok_apel = 6

jeruk_terbeli = 0
anggur_terbeli = 0
apel_terbeli = 0

harga_apel, harga_jeruk, harga_anggur = 10000, 15000, 20000

while apel_terbeli == 0:
    jumlah_beli = int(input("Masukkan jumlah apel : "))
    if jumlah_beli>stok_apel:
        print(f"Jumlah yang di masukan terlalu banyak \n Stock Apel tinggal : {stok_apel}")
    else:
        apel_terbeli = jumlah_beli
        stok_apel -= apel_terbeli
        break
while jeruk_terbeli == 0:
    jumlah_beli = int(input("Masukkan jumlah jeruk : "))
    if jumlah_beli>stok_jeruk:
        print(f"Jumlah yang di masukan terlalu banyak \n Stock Jeruk tinggal : {stok_jeruk}")
    else:
        jeruk_terbeli = jumlah_beli
        stok_jeruk -= jeruk_terbeli
        break
while anggur_terbeli == 0:
    jumlah_beli = int(input("Masukkan jumlah anggur : "))
    if jumlah_beli>stok_anggur:
        print(f"Jumlah yang di masukan terlalu banyak \n Stock Anggur tinggal : {stok_anggur}")
    else:
        anggur_terbeli = jumlah_beli
        stok_anggur -= anggur_terbeli
        break
total_apel = apel_terbeli * harga_apel
total_jeruk = jeruk_terbeli * harga_jeruk
total_anggur = anggur_terbeli * harga_anggur
total_belanja = total_apel + total_jeruk + total_anggur

detail_belanja = f"""
Detail Belanja
Apel: {apel_terbeli} x {harga_apel} = {total_apel}
Jeruk: {jeruk_terbeli} x {harga_jeruk} = {total_jeruk}
Anggur: {anggur_terbeli} x {harga_anggur} = {total_anggur}

Total : {total_belanja}
"""
print("\n\n"+detail_belanja)

uang_masuk = 0
while uang_masuk < total_belanja:
    uang_masuk = int(input("Masukkan jumlah uang : "))
    if uang_masuk < total_belanja:
        print(f"Uang yang anda kurang sebesar : {total_belanja-uang_masuk}")
    else:
        print("Terimakasih")
        if uang_masuk > total_belanja:
            print(f"\nUang kembalian anda : {uang_masuk-total_belanja}")
```
